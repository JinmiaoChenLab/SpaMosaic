```python
from scipy.io import mmread
from scipy.sparse import csr_matrix
import anndata as ad
import pandas as pd
import scvi
import numpy as np
import scanpy as sc
scvi.settings.seed = 420
import os, sys, time, gc
from os.path import join

os.environ['CUDA_VISIBLE_DEVICES'] = '1'
```

```python
data_dir = '../../../data/processed/Misar-E15-E18-E18'

ad_mult_rna = sc.read_h5ad(join(data_dir, 'E15/ad_rna.h5ad'))
ad_mult_atac = sc.read_h5ad(join(data_dir, 'E15/ad_atac.h5ad'))

ad_test_rna = sc.read_h5ad(join(data_dir, 'E18-rna/ad_rna.h5ad'))
ad_test_atac = sc.read_h5ad(join(data_dir, 'E18-atac/ad_atac.h5ad'))
```

```python
ad_rna_all = sc.concat([ad_mult_rna, ad_test_rna])
ad_atac_all = sc.concat([ad_mult_atac, ad_test_atac])

sc.pp.highly_variable_genes(ad_rna_all, flavor='seurat_v3', n_top_genes=10000, batch_key='Sample')
hvg_names = ad_rna_all.var.query('highly_variable').index.to_numpy()

# ac.pp.tfidf(ad_atac_all, scale_factor=1e4)
sc.pp.highly_variable_genes(ad_atac_all, flavor='seurat_v3', n_top_genes=100000, batch_key='Sample')
hvp_names = ad_atac_all.var.query('highly_variable').index.to_numpy()

del ad_rna_all, ad_atac_all
gc.collect()
```

```python
ad_mult_rna = ad_mult_rna[:, hvg_names].copy()
ad_test_rna = ad_test_rna[:, hvg_names].copy()

ad_mult_atac = ad_mult_atac[:, hvp_names].copy()
ad_test_atac = ad_test_atac[:, hvp_names].copy()
```

```python
for adx in [ad_mult_rna, ad_test_rna]:
    adx.var['modality'] = ['Gene Expression']*adx.shape[1]
for adx in [ad_mult_atac, ad_test_atac]:
    adx.var['modality'] = ['Peaks']*adx.shape[1]
```

```python
def colcat(ads):
    adx = sc.concat(ads, axis=1)
    adx.obs = ads[0].obs.copy()
    return adx

def rowcat(ads):
    adx = sc.concat(ads)
    adx.var = ads[0].var.copy()
    return adx
```

```python
ad_mult = colcat([ad_mult_rna, ad_mult_atac])
```

```python
ad_mult.obs['modality'] = 'paired'
ad_test_rna.obs['modality'] = 'rna'
ad_test_atac.obs['modality'] = 'accessibility'
```

```python
# We can now use the organizing method from scvi to concatenate these anndata
adata_mvi = scvi.data.organize_multiome_anndatas(multi_anndata=ad_mult, 
                                                 rna_anndata=ad_test_rna,
                                                 atac_anndata=ad_test_atac)

print(adata_mvi.shape)

adata_mvi = adata_mvi[:, adata_mvi.var["modality"].argsort()].copy()
scvi.model.MULTIVI.setup_anndata(adata_mvi, batch_key="modality", categorical_covariate_keys=['Sample'])
```

```python

model = scvi.model.MULTIVI(
    adata_mvi,
    n_genes=(adata_mvi.var["modality"] == "Gene Expression").sum(),
    n_regions=(adata_mvi.var["modality"] == "Peaks").sum(),
)
model.view_anndata_setup()

model.train()

latent = model.get_latent_representation()

# np.savetxt(arguments[3]+"/MultiVI.csv", latent, delimiter=',')
```
