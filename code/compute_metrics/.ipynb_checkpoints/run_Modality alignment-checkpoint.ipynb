```python
import os, re
import numpy as np
import scanpy as sc
from os.path import join
import pandas as pd

import sys
import scipy.io as sio
import paste as pst
import scipy.sparse as sps
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

sys.path.insert(0, './')
from metrics import FOSCTTM, MS, LabTransfer
```

```python
target_methods = [
    'CLUE',
    'StabMap',
    'SpaMosaic(non-spatial)',
    'UINMF',
    'scMoMaT',
    'SpaMosaic',
    'MIDAS',
    'Cobolt',
    'MultiVI'
]

data_dir = '../../data/processed/'
emb_dir = '../../results/embeddings/'
score_dir = '../../results/scores/'
```

### Mis-E15-E18-E18

```python
ds = 'Misar-E15-E18-E18'

ad1 = sc.read_h5ad(join(data_dir, f'{ds}/E15/ad_rna.h5ad'))
ad2 = sc.read_h5ad(join(data_dir, f'{ds}/E18-rna/ad_rna.h5ad'))
ad3 = sc.read_h5ad(join(data_dir, f'{ds}/E18-atac/ad_atac.h5ad'))
```

```python
df_res = {}
for meth in target_methods:
    if not os.path.exists(join(emb_dir, meth, ds, 'df_emb.csv')):
        continue
    df = pd.read_csv(join(emb_dir, meth, ds, 'df_emb.csv'), index_col=0)
    ad2_tmp = sc.AnnData(df.loc[ad2.obs_names].values, obs=ad2.obs.copy())
    ad3_tmp = sc.AnnData(df.loc[ad3.obs_names].values, obs=ad3.obs.copy())
    ad2_tmp.obsm['emb'] = ad2_tmp.X.copy()
    ad3_tmp.obsm['emb'] = ad3_tmp.X.copy()
    ad2_tmp.obs['batch'] = 'b2'
    ad3_tmp.obs['batch'] = 'b2'
    
    fosc = FOSCTTM(ad2_tmp, ad3_tmp, use_rep='emb')
    ms = MS(ad2_tmp, ad3_tmp, split_by='batch', k=1, use_rep='emb')
    
    df_res[meth] = [fosc, ms]
```

```python
dfm = pd.DataFrame.from_dict(df_res, orient='index', columns=['FOSCTTM', "MS"])
dfm.to_csv(join(score_dir, ds, 'df_modAlign.csv'))
```

### Simulation

```python
ds = 'Simulations'

fds = os.listdir(join(data_dir, ds))
for fd in fds:
    for seed in ['1','2','3']:
        ad2 = sc.read_h5ad(join(data_dir, ds, fd, seed, 'ad_rna2.h5ad'))
        ad3 = sc.read_h5ad(join(data_dir, ds, fd, seed, 'ad_adt2.h5ad'))
        
        df_res = {}
        for meth in target_methods:
            if not os.path.exists(join(emb_dir, meth, ds, fd, seed, 'df_emb.csv')):
                continue
            df = pd.read_csv(join(emb_dir, meth, ds, fd, seed, 'df_emb.csv'), index_col=0)
            ad2_tmp = sc.AnnData(df.loc[ad2.obs_names].values, obs=ad2.obs.copy())
            ad3_tmp = sc.AnnData(df.loc[ad3.obs_names].values, obs=ad3.obs.copy())
            ad2_tmp.obsm['emb'] = ad2_tmp.X.copy()
            ad3_tmp.obsm['emb'] = ad3_tmp.X.copy()
            ad2_tmp.obs['batch'] = 'b2'
            ad3_tmp.obs['batch'] = 'b2'

            df_res[meth] = LabTransfer(ad2_tmp, ad3_tmp, 'emb', 'gt', knn=10)
            
        df_res = pd.DataFrame.from_dict(df_res, orient='index', columns=['LabTransfer'])
        
        os.makedirs(join(score_dir, ds, fd, seed), exist_ok=True)
        df_res.to_csv(join(score_dir, ds, fd, seed, 'df_labTransfer.csv'))
```
